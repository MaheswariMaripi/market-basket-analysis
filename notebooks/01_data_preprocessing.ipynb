# 01 - Data Preprocessing

In this notebook we load the raw market basket dataset, handle
missing values and convert it into a clean transaction format
ready for association rule mining.

The reusable functions live in `src/preprocessing.py`.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
from src.preprocessing import load_raw_dataframe

raw = load_raw_dataframe()
print('Shape of the raw dataset:', raw.shape)
raw.head()
```

```python
print('Missing values per column:')
print(raw.isna().sum().describe().astype(int))
print(f'Total empty cells: {raw.isna().sum().sum():,}')
```

```python
from src.preprocessing import clean_transactions

transactions = clean_transactions(raw)
print(f'Number of non-empty transactions: {len(transactions):,}')
transactions[:5]
```

```python
from src.preprocessing import one_hot_encode

encoded = one_hot_encode(transactions)
print('One-hot encoded shape (transactions x products):', encoded.shape)
encoded.head()
```

## Summary

* The raw file contains sparse, ragged rows with many empty cells.
* Empty cells, whitespace and duplicate items were removed.
* The data is now a binary matrix: 1 means the product appears in
  the transaction, 0 means it does not - the exact input format
  expected by the Apriori algorithm.
