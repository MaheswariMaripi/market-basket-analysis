## Observations

* A small group of products (e.g. mineral water, eggs, milk) is
  bought very frequently while most products appear rarely.
* Most baskets contain between 1 and a handful of items - the
  long tail is typical for market basket data.
* The right skew confirms that simple frequency ranking alone is
  not enough; association rules are needed to find combinations.
