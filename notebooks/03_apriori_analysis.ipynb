# 03 - Frequent Itemsets & Association Rules (Apriori)

We apply the Apriori algorithm to find frequently co-occurring
products and then extract association rules with support,
confidence and lift.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
from src.preprocessing import preprocess_dataset

raw, transactions, encoded = preprocess_dataset()
encoded.head()
```

### Mining frequent itemsets

`min_support=0.003` means an itemset must appear in at least
`0.003 * 7501 ≈ 22` transactions to be considered frequent.

```python
from src.association_rules import mine_frequent_itemsets, generate_rules

itemsets = mine_frequent_itemsets(encoded, min_support=0.003, max_len=3)
print(f'Frequent itemsets found: {len(itemsets):,}')
itemsets.sort_values('support', ascending=False).head(10)
```

```python
rules = generate_rules(itemsets, min_confidence=0.2, min_lift=1.5)
print(f'Association rules found: {len(rules):,}')
rules.head(10)
```

### Strongest rules by lift

Lift > 1 means the products are positively associated: buying the
antecedent makes the consequent more likely.

```python
from src.association_rules import top_rules, plot_top_rules

top_rules(rules, n=10)
```

```python
plot_top_rules(rules, top_n=20)
```
